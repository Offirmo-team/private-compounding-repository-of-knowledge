### Random user name



```typescript
Math.random().toString(36) // base-36 = digits 0-9 plus letters a-z PLUS a 'O.' prefix
```

```typescript
let length = 10
Math.random().toString(36).slice(2).padEnd(length, '0').slice(0, length);
```
